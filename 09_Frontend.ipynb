# Phase 9 — Frontend

## Objective

The goal of this phase is to build a user-facing interface for the
Missing Person Finder application.

The frontend will allow a user to:

- Upload a photograph
- Submit the photograph for face search
- Communicate with the FastAPI backend
- Display the search status
- Display ranked potential matches
- View relevant missing-person information
- Support human verification of potential candidates

The frontend will communicate with the backend through the `/search`
API endpoint developed in Phase 8.
